```python
import requests
import matplotlib.pyplot as plt 
```

```python
base_url = "https://datahub-v2-s3.arlingtonva.us/Uploads/AutomatedJobs/Google+Analytics.json?v=1792594207"

# returns the top 10000 records
url_filters = "?$top=10"

full_url = base_url + url_filters

response = requests.get(full_url)
meoww = response.json()
#print(data[:30])
print(meoww[:30])
```

```python
url_list_raw = [item["Page"] for item in meoww if "Page" in item]
print(url_list_raw)
```

```python

# removes duplicates 
url_list_clean = list(set(url_list_raw))
# Puts list in alphibetical order
url_list_clean.sort()
print(*url_list_clean, sep="\n")

# print(*url_list_raw, sep="\n")
# prints url_list_raw with one url per line
```
