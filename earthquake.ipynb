```python
%pip install requests

import requests
import matplotlib.pyplot as plt 
```

```python
# build url 

base_url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

full_url = base_url + "?format=csv"

### Search
full_url = (full_url
            + "&starttime=20101030"
            + "&endtime=2010-10-31"
            + "&latitude=38.8"
            + "&longitude=-77"
            + "&maxradiuskm=1000"
            )

print(full_url)
```

```python
response = requests.get(full_url)
```

```python
response.text.splitlines()[0:10]
```

## Reproduiceing excel functions

```python
# how many earthquake?

headers = response.text.splitlines()[0]
records = response.text.splitlines()[1:]

len(records)
```

```python
# Highest magnitued
magnitudes = []

for r in records:
    try:
        magnitudes.append(float(r.split(",")[4]))
    except:
        pass

max(magnitudes)
```

## Visualise

```python
plt.hist(magnitudes)
plt.show
```
