```python
import numpy as np
import matplotlib.pyplot as plt 
from matplotlib.colors import rgb_to_hsv, hsv_to_rgb
```

```python
img = plt.imread('GHC.png')
img = img[:, :, :3]

#convert to HSV (hue, satiration value)
hsv = rgb_to_hsv(img)
```

```python
h = hsv[:, :, 0]
s = hsv[:, :, 1]
v = hsv[:, :, 2]

print(hsv.shape, hsv.dtype)
print('hue range:', h.min(), h.max())
print('saturation range:', s.min(), s.max())
print('value range:', v.min(), v.max())
```

```python
# FILTERS 

def burger_and_fried():
    new_hsv[:, :, 0] = (new_hsv[:, :, 0] - 0.05) % 1.0
    new_hsv[:, :, 1] = (new_hsv[:, :, 1] + 0.45) % 1.0
    new_hsv[:, :, 2] = (new_hsv[:, :, 2] - 0.1) % 1.0
```

```python
# Make your changes on a copy, so you can compare to the original
new_hsv = hsv.copy()

burger_and_fried()

# Example filter
# new_hsv[:, :, 2] = (new_hsv[:, :, 2] + 0.33) % 1.0

# All values must stay between 0 and 1 before converting back
new_hsv = np.clip(new_hsv, 0, 1)

# Convert HSV -> RGB
new_img = hsv_to_rgb(new_hsv)
```

```python
# render the original and the result side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(img)
ax1.set_title('Original')
ax2.imshow(new_img)
ax2.set_title('Filtered')
plt.show()
```

```python
# new image 
plt.imsave('filtered.png', new_img)
```
